## Data Preprocesing

```python
print("Jumlah data duplikat:", df.duplicated().sum())
```

```python
df[df.duplicated(keep=False)].sort_values(by=df.columns.tolist())
```

```python
df = df.drop_duplicates().copy()
```

```python
print("Jumlah data setelah handling duplicate:", len(df))
print("Jumlah duplicate setelah handling:", df.duplicated().sum())
```

```python
missing_value = df.isnull().sum()

print("Jumlah missing value setiap kolom:")
display(missing_value)
```

```python
usia_tidak_masuk_akal = df[
    (df["Usia"] < 17) | (df["Usia"] > 100)
]

print("Jumlah usia tidak masuk akal:", len(usia_tidak_masuk_akal))
display(usia_tidak_masuk_akal[["ID_Pelanggan", "Usia"]])
```

```python
rating_tidak_masuk_akal = df[
    (df["Rating"] < 1) | (df["Rating"] > 5)
]

print("Jumlah rating tidak masuk akal:", len(rating_tidak_masuk_akal))
display(rating_tidak_masuk_akal[["ID_Pelanggan", "Rating"]])
```

```python
diskon_tidak_masuk_akal = df[
    (df["Diskon_Persen"] < 0) | (df["Diskon_Persen"] > 100)
]

print("Jumlah diskon tidak masuk akal:", len(diskon_tidak_masuk_akal))
display(diskon_tidak_masuk_akal[["ID_Pelanggan", "Diskon_Persen"]])
```

```python
print("Jumlah tidak masuk akal:")

print("Pendapatan <= 0 :", (df["Pendapatan"] <= 0).sum())
print("Harga_Satuan <= 0 :", (df["Harga_Satuan"] <= 0).sum())
print("Jumlah <= 0 :", (df["Jumlah"] <= 0).sum())
print("Lama_Pengiriman_Hari <= 0 :", (df["Lama_Pengiriman_Hari"] <= 0).sum())
print("Total_Pembelian <= 0 :", (df["Total_Pembelian"] <= 0).sum())
```
